```python
import numpy as np
```

```python
#定義激活函數
def relu(x):
    return np.maximum(0, x)

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def softmax(x):
    c=np.max(x)
    x = x - c
    return np.exp(x) / np.sum(np.exp(x))

def cross_entropy_error(y_hat, y):
    return -np.sum(y * np.log(y_hat + 1e-7))
```

```python
#輸入層:六個神經元
x = np.array([12, 0.5, 2.5, 8, 9, 2])
print(f"輸入資料x:{x},形狀:{x.shape}")
#第一隱藏層:五個神經元
W1 = np.array([
    [ 0.1, -0.2,  0.3,  0.1,  0.2],
    [-0.1,  0.4,  0.2, -0.3,  0.1],
    [ 0.3,  0.1, -0.2,  0.5,  0.4],
    [ 0.2, -0.4,  0.1,  0.2, -0.1],
    [-0.3,  0.2,  0.5, -0.1,  0.3],
    [ 0.1,  0.3, -0.1,  0.4,  0.2]
    ])
b1 = np.array([0.1, -0.1, 0.2, 0.0, 0.3])
print(f"第一隱藏層W1形狀:{W1.shape},b1形狀:{b1.shape},參數個數:{W1.size + b1.size}")
#第二隱藏層:四個神經元
W2 = np.array([
    [ 0.2,  0.4, -0.1,  0.3],
    [-0.3,  0.1,  0.5, -0.2],
    [ 0.1, -0.2,  0.3,  0.4],
    [ 0.4,  0.3, -0.2,  0.1],
    [-0.1,  0.5,  0.1, -0.3]
    ])
b2 = np.array([-0.2, 0.1, 0.3, -0.1])
print(f"第二隱藏層W2形狀:{W2.shape}, b2形狀:{b2.shape}, 參數個數:{W2.size + b2.size}")
#第三隱藏層:四個神經元
W3 = np.array([
    [ 0.3, -0.1,  0.2,  0.4],
    [ 0.1,  0.5, -0.3,  0.1],
    [-0.4,  0.2,  0.1, -0.2],
    [ 0.2, -0.3,  0.4,  0.3]
    ])
b3 = np.array([0.1, 0.2, -0.1, 0.0])
print(f"第三隱藏層W3形狀:{W3.shape}, b3形狀:{b3.shape}, 參數個數:{W3.size + b3.size}")
#輸出層:三個神經元
W4 = np.array([
    [ 0.5,  0.2, -0.3],
    [-0.1,  0.4,  0.1],
    [ 0.3, -0.2,  0.6],
    [ 0.2,  0.1, -0.4]
    ])
b4 = np.array([0.1, -0.1, 0.2])
print(f"輸出層W4形狀:{W4.shape}, b4形狀:{b4.shape}, 參數個數:{W4.size + b4.size}")
```

```python
#第一隱藏層relu
z1 = np.dot(x, W1) + b1
h1 = relu(z1)
print(f"第一層z1:{z1}, 形狀:{z1.shape}")
print(f"第一層 h1:{h1}, 形狀:{h1.shape}")
#第二隱藏層relu
z2 = np.dot(h1, W2) + b2
h2 = relu(z2)
print(f"第二層z2:{z2}, 形狀:{z2.shape}")
print(f"第二層h2:{h2}, 形狀:{h2.shape}")
#第三隱藏層sigmoid
z3 = np.dot(h2, W3) + b3
h3 = sigmoid(z3)
print(f"第三層z3: {z3}, 形狀:{z3.shape}")
print(f"第三層h3: {h3}, 形狀:{h3.shape}")
#輸出層Softmax
z4 = np.dot(h3, W4) + b4
y_hat = softmax(z4)
print(f"輸出層z4:{z4}, 形狀:{z4.shape}")
print(f"最終輸出y_hat:{y_hat}, 形狀:{y_hat.shape}")
# 4.cross_entropy_error
y = np.array([0, 0, 1])
loss = cross_entropy_error(y_hat, y)
print(f"真實結果y:{y}")
print(f"與y_hat之間的cross_entropy_error:{loss}")
```

```python
import gradio as gr
```

```python
def dnn(x):
    x = np.array(x, dtype=float)
    #第一層
    z1 = np.dot(x, W1) + b1
    h1 = relu(z1)
    #第二層
    z2 = np.dot(h1, W2) + b2
    h2 = relu(z2)
    #第三層
    z3 = np.dot(h2, W3) + b3
    h3 = sigmoid(z3)
    #輸出層
    z4 = np.dot(h3, W4) + b4
    y_hat = softmax(z4)
    return y_hat

xtest = np.array([12, 0.5, 2.5, 8, 9, 2])
y_hat = dnn(xtest)

print(f"輸入xtest時,dnn函數輸出的y_hat為：{y_hat}")

#互動頁面
def gradio_interface(x1,x2,x3,x4,x5,x6):
    x_input = np.array([x1,x2,x3,x4,x5,x6])
    y_hat = dnn(x_input)
    highest = int(np.argmax(y_hat) + 1)
    return f"第{highest}類", f"類別1:{y_hat[0]:.5f}\n類別2:{y_hat[1]:.5f}\n類別3:{y_hat[2]:.5f}"

interface = gr.Interface(
    fn=gradio_interface,
    inputs=[
        gr.Number(value=12.0, label="特徵 1"),
        gr.Number(value=0.5, label="特徵 2"),
        gr.Number(value=2.5, label="特徵 3"),
        gr.Number(value=8.0, label="特徵 4"),
        gr.Number(value=9.0, label="特徵 5"),
        gr.Number(value=2.0, label="特徵 6")
    ],
    outputs=[
        gr.Textbox(label="機率最高的類別(1,2,或3)"),
        gr.Textbox(label="各類別機率(y_hat)")
    ],
    description="請輸入6個特徵值:。"
)

interface.launch()
```
